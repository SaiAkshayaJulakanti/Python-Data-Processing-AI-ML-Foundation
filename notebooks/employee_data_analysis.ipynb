## 1️⃣ NumPy Foundations & Broadcasting Demonstrations (20 Marks)

In this section, we explore multi-dimensional array operations, statistical aggregations, boolean masking, and broadcasting rules.

```python
# 1. Array Creation & Attribute Inspection
arr_1d = np.array([10, 20, 30, 40, 50])
arr_2d = np.array([[10, 20, 30], [40, 50, 60]])
arr_zeros = np.zeros((3, 4))
arr_linspace = np.linspace(0, 1, 5)

print(f"1D Array: {arr_1d} | Shape: {arr_1d.shape} | Dtype: {arr_1d.dtype}")
print(f"2D Array (2x3):\n{arr_2d}\nShape: {arr_2d.shape} | Dimensions: {arr_2d.ndim}")
print(f"Linspace (0 to 1, 5 points): {arr_linspace}")

# 2. Mathematical Operations & Statistical Aggregations
salaries = np.array([65000, 85000, 110000, 55000, 95000, 125000])
tax_deductions = salaries * 0.20  # Element-wise multiplication
net_salaries = salaries - tax_deductions

print(f"\nBase Salaries: {salaries}")
print(f"Net Salaries after 20% tax: {net_salaries}")
print(f"Total Payroll: ${np.sum(salaries):,.2f}")
print(f"Mean Salary: ${np.mean(salaries):,.2f} | Std Dev: ${np.std(salaries):,.2f}")

# 3. Indexing, Slicing & Boolean Masking
high_earners = salaries[salaries > 90000]
print(f"High Earners (> $90,000): {high_earners}")
```

```python
# 4. Understanding & Applying NumPy Broadcasting Rules
# 4 Employees x 3 Quarters Performance Matrix (Shape: 4, 3)
performance = np.array([
    [85, 90, 92],
    [70, 75, 80],
    [95, 98, 96],
    [60, 65, 70]
])

# Department quarterly weight multipliers (Shape: 3,)
quarter_weights = np.array([1.05, 1.10, 1.15])

# Broadcasting (4, 3) * (3,)
weighted_performance = performance * quarter_weights

print("Original Performance Score Matrix (4x3):\n", performance)
print("\nQuarter Weight Vector (1x3):", quarter_weights)
print("\nBroadcasting Result (Weighted Scores 4x3):\n", np.round(weighted_performance, 2))
```

## 2️⃣ Pandas Data Handling & Merging (30 Marks)

Demonstrating DataFrame creation, multi-column sorting, filtering with `query()`, grouping, and merging.

```python
# 1. Create DataFrame from Dictionary
emp_dict = {
    'Emp_ID': [201, 202, 203, 204, 205],
    'Name': ['Alice Johnson', 'Bob Smith', 'Charlie Lee', 'Diana Prince', 'Evan Wright'],
    'Department': ['Engineering', 'Marketing', 'Engineering', 'Sales', 'Marketing'],
    'Salary': [92000, 68000, 105000, 74000, 62000],
    'Hire_Year': [2018, 2020, 2016, 2021, 2019]
}
df_sample = pd.DataFrame(emp_dict)

# 2. Multi-column Filtering & Sorting
filtered_df = df_sample.query("Salary > 70000").sort_values(by=['Department', 'Salary'], ascending=[True, False])
    print("Filtered & Sorted DataFrame:\n", filtered_df)

# 3. Merging DataFrames
dept_info = pd.DataFrame({
    'Department': ['Engineering', 'Marketing', 'Sales'],
    'Building': ['Building Alpha', 'Building Beta', 'Building Gamma'],
    'Manager': ['Dr. Aris', 'Elena Vance', 'Marcus Vance']
})

merged_df = pd.merge(df_sample, dept_info, on='Department', how='inner')
print("\nMerged DataFrame:\n", merged_df[['Emp_ID', 'Name', 'Department', 'Building', 'Manager']])
```

## 3️⃣ Practical Assignment: Employee Data Cleaning Pipeline (50 Marks)

Here we execute the end-to-end data processing workflow on `data/employees.csv`:
`CSV -> Pandas DataFrame -> Data Cleaning -> Missing Value Handling -> Duplicate Removal -> Filtering -> Grouping -> Statistics -> Clean Dataset`

```python
# 1. Load Raw CSV and Inspect Initial Anomalies
raw_csv_path = '../data/employees.csv'
clean_csv_path = '../data/cleaned_employees.csv'

raw_df = pd.read_csv(raw_csv_path)
print(f"Raw Dataset Loaded: {len(raw_df)} rows, {len(raw_df.columns)} columns")
print("\nMissing Values Summary in Raw Data:\n", raw_df.isna().sum())
print(f"\nDuplicate Rows in Raw Data: {raw_df.duplicated().sum()}")
```

```python
# 2. Instantiate and Run Data Cleaning Pipeline
cleaner = EmployeeDataCleaner(raw_csv_path)
cleaned_df, stats, logs = cleaner.run_pipeline(clean_csv_path)

print("\n=== PIPELINE EXECUTION LOGS ===")
for log in logs:
    print(f"• {log}")
```

```python
# 3. Verify Clean Dataset
print(f"\nClean Dataset Saved to: {clean_csv_path}")
print(f"Clean Dataset Records: {len(cleaned_df)}")
print("Remaining Missing Values:\n", cleaned_df.isna().sum())
print("\nFirst 5 Records of Clean Dataset:\n")
print(tabulate(cleaned_df.head(), headers='keys', tablefmt='psql'))
```

```python
# 4. Departmental Group Statistics & Aggregations
dept_stats = stats['department_statistics']
print("=== DEPARTMENTAL SALARY & EXPERIENCE AGGREGATIONS ===")
print(tabulate(dept_stats, headers='keys', tablefmt='psql'))

print("\n=== OVERALL COMPANY METRICS ===")
for k, v in stats['overall_statistics'].items():
    print(f"• {k}: {v:,.2f}" if isinstance(v, float) else f"• {k}: {v}")
```
